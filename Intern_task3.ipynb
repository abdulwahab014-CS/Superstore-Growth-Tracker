# Convert dates into proper date format

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%m/%d/%Y",
    errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    format="%m/%d/%Y",
    errors="coerce"
)

# Confirms that both date columns were successfully converted.

df[["Order Date", "Ship Date"]].dtypes
```

# Standardize column names

1.   List item
2.   List item



```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)

df.columns
```

# Clean text values

```python
text_columns = df.select_dtypes(
    include="object"
).columns

for col in text_columns:
    df[col] = (
        df[col]
        .str.replace("\xa0", " ", regex=False)
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )
```

# Remove duplicate rows

```python
before = len(df)

df = df.drop_duplicates().copy()

after = len(df)

print("Rows before cleaning:", before)
print("Rows after cleaning:", after)
print("Duplicates removed:", before - after)
```

# Check invalid values

# Handle missing values

```python
# Checks missing values before handling them.

print(df.isnull().sum())

# Removes records missing essential order information.

df = df.dropna(
    subset=["order_id", "order_date", "sales"]
).copy()

# Fills missing categorical values with Unknown.

text_columns = df.select_dtypes(
    include="object"
).columns

for col in text_columns:
    df[col] = df[col].fillna("Unknown")

# Fills missing numerical values with the median.

numeric_columns = [
    "sales", "quantity", "discount", "profit"
]

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

print("Missing values after cleaning:")
print(df.isnull().sum())
```

# Detect outliers using IQR

```python
# Identifies unusually high or low numerical values using the IQR method.

numeric_cols = [
    "sales", "quantity", "discount", "profit"
]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    print(f"{col}: {outliers} potential outliers")
```

# Create date-based features

```python
# Creates year, month, quarter, and year-month features
# to analyze sales trends over time.

df["order_year"] = df["order_date"].dt.year

df["order_month"] = df["order_date"].dt.month

df["month_name"] = df["order_date"].dt.month_name()

df["order_quarter"] = df["order_date"].dt.quarter

df["year_month"] = (
    df["order_date"].dt.to_period("M").astype(str)
)

df["day_of_week"] = df["order_date"].dt.day_name()

df.head()
```

# Calculate shipping duration

```python
# Calculates the number of days between ordering and shipping.

df["shipping_days"] = (
    df["ship_date"] - df["order_date"]
).dt.days

df[[
    "order_date",
    "ship_date",
    "shipping_days"
]].head()
```

# Calculate profit margin and sales per unit

```python
# Calculates profit margin as a percentage of sales
# and the average sales value per unit.

df["profit_margin_pct"] = np.where(
    df["sales"] != 0,
    (df["profit"] / df["sales"]) * 100,
    np.nan
)

df["sales_per_unit"] = (
    df["sales"] / df["quantity"]
)

df[[
    "sales",
    "profit",
    "quantity",
    "profit_margin_pct",
    "sales_per_unit"
]].head()
```

# Create profit status

```python
# Classifies each transaction as Profit or Loss.

df["profit_status"] = np.where(
    df["profit"] >= 0,
    "Profit",
    "Loss"
)

df["profit_status"].value_counts()
```

# Create discount groups

```python
# Groups transactions into discount categories
# to compare the impact of discounts on sales and profit.

df["discount_band"] = pd.cut(
    df["discount"],
    bins=[-0.01, 0, 0.2, 0.5, 1.0],
    labels=[
        "No Discount",
        "Low",
        "Medium",
        "High"
    ]
)
df["discount_band"].value_counts()
```

# Check newly created features

# Calculate main business KPIs

```python
# Calculates the main business performance indicators.

total_sales = df["sales"].sum()

total_orders = df["order_id"].nunique()

total_customers = df["customer_id"].nunique()

total_products = df["product_id"].nunique()

total_quantity = df["quantity"].sum()

total_profit = df["profit"].sum()

average_order_value = total_sales / total_orders

profit_margin = (
    total_profit / total_sales
) * 100

kpi = pd.DataFrame({
    "KPI": [
        "Total Sales",
        "Total Orders",
        "Total Customers",
        "Total Products",
        "Total Quantity Sold",
        "Total Profit",
        "Average Order Value",
        "Profit Margin (%)"
    ],
    "Value": [
        total_sales,
        total_orders,
        total_customers,
        total_products,
        total_quantity,
        total_profit,
        average_order_value,
        profit_margin
    ]
})

display(kpi)
```

# Monthly revenue

```python
# Calculates monthly revenue, profit, and order counts.

monthly_revenue = (
    df.groupby("year_month")
    .agg(
        revenue=("sales", "sum"),
        profit=("profit", "sum"),
        orders=("order_id", "nunique")
    )
    .reset_index()
)

monthly_revenue["revenue_growth_pct"] = (
    monthly_revenue["revenue"].pct_change() * 100
)

display(monthly_revenue)
```

# Top 10 products

```python
# Finds the ten products with the highest total sales.

top_products = (
    df.groupby(
        ["product_id", "product_name"]
    )
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        quantity_sold=("quantity", "sum")
    )
    .reset_index()
    .sort_values(
        "total_sales",
        ascending=False
    )
    .head(10)
)

display(top_products)
```

#Top 10 customers

```python
# Finds the ten customers with the highest total sales.

top_customers = (
    df.groupby(
        ["customer_id", "customer_name"]
    )
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        total_orders=("order_id", "nunique")
    )
    .reset_index()
    .sort_values(
        "total_sales",
        ascending=False
    )
    .head(10)
)

display(top_customers)
```

# Regional performance

```python
# Compares sales, profit, orders, and profit margins by region.

regional_performance = (
    df.groupby("region")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        total_orders=("order_id", "nunique")
    )
    .reset_index()
)

regional_performance["profit_margin_pct"] = (
    regional_performance["total_profit"] /
    regional_performance["total_sales"] * 100
)

regional_performance = regional_performance.sort_values(
    "total_sales",
    ascending=False
)

display(regional_performance)
```

#Category comparison

```python
# Compares the performance of Furniture,
# Office Supplies, and Technology.

category_comparison = (
    df.groupby("category")
    .agg(
        total_sales=("sales", "sum"),
        total_profit=("profit", "sum"),
        quantity_sold=("quantity", "sum"),
        total_orders=("order_id", "nunique")
    )
    .reset_index()
)

category_comparison["profit_margin_pct"] = (
    category_comparison["total_profit"] /
    category_comparison["total_sales"] * 100
)

category_comparison["sales_share_pct"] = (
    category_comparison["total_sales"] /
    category_comparison["total_sales"].sum() * 100
)

display(category_comparison)
```

# Profit analysis

```python
# Identifies profitable and loss-making transactions.

profit_analysis = pd.DataFrame({
    "Metric": [
        "Total Profit",
        "Profitable Transactions",
        "Loss-Making Transactions",
        "Average Profit per Transaction",
        "Largest Profit",
        "Largest Loss"
    ],
    "Value": [
        df["profit"].sum(),
        (df["profit"] > 0).sum(),
        (df["profit"] < 0).sum(),
        df["profit"].mean(),
        df["profit"].max(),
        df["profit"].min()
    ]
})

display(profit_analysis)
```

```python
# 1. Total Sales and headline KPIs
kpis = pd.DataFrame([{
    "total_sales":df.sales.sum(), "total_profit":df.profit.sum(),
    "total_orders":df.order_id.nunique(), "total_customers":df.customer_id.nunique(),
    "total_quantity":df.quantity.sum(),
    "profit_margin_pct":df.profit.sum()/df.sales.sum()*100 if df.sales.sum() else 0
}])

# 2. Monthly Revenue
monthly = df.groupby(["order_year","order_month","month_name","year_month"],as_index=False).agg(
    monthly_revenue=("sales","sum"),monthly_profit=("profit","sum"),
    units_sold=("quantity","sum"),total_orders=("order_id","nunique")
).sort_values(["order_year","order_month"])

# 3. Top Products
pg=["product_id","product_name"]+[c for c in ["category","sub_category"] if c in df]
products=df.groupby(pg,as_index=False,dropna=False).agg(
    units_sold=("quantity","sum"),total_sales=("sales","sum"),total_profit=("profit","sum")
).nlargest(10,"total_sales")

# 4. Top Customers
cg=["customer_id","customer_name"]+[c for c in ["segment"] if c in df]
customers=df.groupby(cg,as_index=False,dropna=False).agg(
    total_sales=("sales","sum"),total_profit=("profit","sum"),total_orders=("order_id","nunique")
).nlargest(10,"total_sales")

# 5. Regional Performance
regions=df.groupby("region",as_index=False,dropna=False).agg(
    total_sales=("sales","sum"),total_profit=("profit","sum"),
    total_quantity=("quantity","sum"),total_orders=("order_id","nunique")
).sort_values("total_sales",ascending=False)
regions["profit_margin_pct"]=np.where(regions.total_sales!=0,regions.total_profit/regions.total_sales*100,0)

# 6. Profit Analysis
profit=df.groupby("profit_status",as_index=False,dropna=False).agg(
    transaction_rows=("profit","size"),total_sales=("sales","sum"),
    total_profit=("profit","sum"),total_quantity=("quantity","sum")
)

# 7. Sales Trend
trend=df.groupby("order_year",as_index=False).agg(
    total_sales=("sales","sum"),total_profit=("profit","sum"),
    total_quantity=("quantity","sum"),total_orders=("order_id","nunique")
).sort_values("order_year")

# 8. Category Comparison
categories=df.groupby("category",as_index=False,dropna=False).agg(
    total_sales=("sales","sum"),total_profit=("profit","sum"),
    total_quantity=("quantity","sum"),total_orders=("order_id","nunique")
).sort_values("total_sales",ascending=False)
categories["profit_margin_pct"]=np.where(categories.total_sales!=0,categories.total_profit/categories.total_sales*100,0)
```
